```python
exclude_cols = merged_df.select_dtypes(exclude=['object', 'datetime', 'datetime64[ns]']).columns

for col in merged_df.columns:
    if col in exclude_cols:
        continue

    print(f"\n📌 컬럼명: {col}")
    desc = merged_df[col].describe()

    # 포맷팅 함수: 천 단위 쉼표 + 소수점 2자리
    def format_val(x):
        try:
            if pd.isnull(x):
                return "NaN"
            elif abs(x) >= 1:
                return f"{x:,.2f}"
            else:
                return f"{x:.2f}"
        except:
            return str(x)

    # 포맷된 출력
    for stat_name in desc.index:
        value = desc[stat_name]
        print(f"{stat_name:>10}: {format_val(value)}")
    
    print("-" * 50)
```

```python
exclude_cols = merged_df.select_dtypes(exclude=['int64', 'float64']).columns

for col in merged_df.columns:
    if col in exclude_cols:
        continue

    print(f"\n📌 컬럼명: {col}")
    desc = merged_df[col].describe()

    # 포맷팅 함수: 천 단위 쉼표 + 소수점 2자리
    def format_val(x):
        try:
            if pd.isnull(x):
                return "NaN"
            elif abs(x) >= 1:
                return f"{x:,.2f}"
            else:
                return f"{x:.2f}"
        except:
            return str(x)

    # 포맷된 출력
    for stat_name in desc.index:
        value = desc[stat_name]
        print(f"{stat_name:>10}: {format_val(value)}")

    # 0 값 개수 추가
    zero_count = (merged_df[col] == 0).sum()
    print(f"{'Zero count':>10}: {format_val(zero_count)}")

    print("-" * 50)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# 2. 숫자형 컬럼만 필터링
df_numeric = merged_df.select_dtypes(include=["number"])

# 3. 상관계수 계산
corr_matrix = df_numeric.corr(method="pearson")  # spearman 또는 kendall도 가능

# 4. 상관계수 시각화 (heatmap)
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("📊 변수 간 상관관계")
plt.tight_layout()
plt.show()
```
