```python
def get_team_stats_from_markup(markup):
    soup = BeautifulSoup(markup, 'html.parser')
    table = soup.find('table')
    df = pd.read_html(str(table))[0]
    pids = []
    trs = soup.find_all('tr')
    for tr in trs:
        a_tag = tr.find('a')
        if a_tag:
            pid = a_tag['href'][-5:]
            pids.append(pid)
        else:
            pids.append(None)
    return df,pids
```
